### 1) Data loading

```python
import sys
sys.path.append('..')
from data_loader import get_data

X_train, X_val, X_test, y_train, y_val, y_test = get_data()
print('X_tr:', X_train.shape, 'X_te:', X_test.shape)
```

### 2) Training

##### Iteration 1

```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

# Model
mlp = MLPClassifier(
    hidden_layer_sizes=(256, 128),
    activation='relu',
    learning_rate_init=0.001,
    max_iter=50,
    random_state=42
)

mlp.fit(X_train, y_train)

val_preds = mlp.predict(X_val)
val_accuracy = accuracy_score(y_val, val_preds)
print(f"Validation accuracy: {val_accuracy:.4f}")
print(f"Validation error rate: {1 - val_accuracy:.4f}")

# Plot
plt.plot(mlp.loss_curve_)
plt.title("Training Loss Over Iterations")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.show()

# This reached the maximum iterations and did not converge
```

### 3) Testing

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

mlp = MLPClassifier(
    hidden_layer_sizes=(256, 128),
    activation='relu',
    learning_rate_init=0.001,
    max_iter=50,
    random_state=42
)

mlp.fit(X_train, y_train)

test_preds = mlp.predict(X_test)
test_accuracy = accuracy_score(y_test, test_preds)
print(f"Test accuracy: {test_accuracy:.4f}")
print(f"Test error rate: {1 - test_accuracy: .4f}")

class_names = ['T-shirt', 'Trouser', 'Pullover', 'Dress', 'Coat', 
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Boot']

cm = confusion_matrix(y_test, test_preds)
disp = ConfusionMatrixDisplay(cm, display_labels=class_names)

fig, ax = plt.subplots(figsize=(10, 8))
disp.plot(ax=ax)
plt.title("FNN Confusion Matrix")
plt.tight_layout()
plt.show()
```
